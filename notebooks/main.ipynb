```python
from IPython.core.display import Markdown
from IPython.core.display_functions import display
import pandas as pd
from pathlib import Path

""" métodos de verificación y carga de archivos """
def verificar_archivo(str) -> Path:
    # Archivo de encuesta de inversores
    ruta = Path("./csv/") / str
    
    # Verificar exitencia
    if ruta.exists():
            return ruta
            
    # Devolver error Archivo no encontrado
    raise FileNotFoundError(
        "No se encontró el directorio con los archivos del Proyecto C."
    )

def cargar_datos(archivos):
    dict_datos = {}
    for nombre, nombre_archivo in archivos.items():
        # verificar que cada archivo existe
        ruta = verificar_archivo(nombre_archivo)
        # cargar datos
        dict_datos[nombre] = pd.read_csv(ruta)

    return dict_datos

""" métodos de análisis de datos """
def mostrar_primeras_filas(nombre: str, datos: pd.DataFrame) -> None:
    # 1.1 mostrar las primeras filas 
    display(Markdown(f"### **Archivo: {nombre}.csv**"))
    display(datos.head())

def mostrar_descripcion(datos: pd.DataFrame) -> None:
    # 1.2
    # cantidad de filas y columnas
    filas, columnas = datos.shape
    display(Markdown(f"**Filas y columnas:** {filas} filas, {columnas} columnas"))

    # tipos de dato
    display(Markdown("#### Tipos de dato por columna"))
    display(datos.dtypes.to_frame(name="tipo_dato").T)

    # cantidad de categorías en las variables categóricas
    categoricas = datos.select_dtypes(include=["object", "category", "string", "bool"])
    if not categoricas.empty:
        display(Markdown("#### Cantidad de categorías (variables categóricas)"))
        display(categoricas.nunique().to_frame(name="cantidad_categorias"))

    # descripción estadística de las continuas
    continuas = datos.select_dtypes(include="number")
    if not continuas.empty:
        display(Markdown("#### Descripción estadística (variables continuas)"))
        display(continuas.describe().T)
    return

def mostrar_duplicados(datos: pd.DataFrame) -> None:
    # 1.3 búsqueda de formularios enviados dos veces
    # y operaciones por duplicado (solo se muestran, no se eliminan)
    # duplicados exactos (todas las columnas iguales)
    exactos = datos.duplicated()
    display(Markdown(
        f"#### Duplicados exactos: {exactos.sum()} filas "
    ))
    if exactos.any():
        display(datos[datos.duplicated(keep=False)].sort_values(list(datos.columns)).head(10))

    # formularios enviados dos veces: mismo id_cliente en la encuesta
    if "id_cliente" in datos.columns and "id_movimiento" not in datos.columns:
        repetidos = datos[datos.duplicated(subset="id_cliente", keep=False)]
        display(Markdown(
            f"#### Formularios con id_cliente repetido: "
            f"{datos.duplicated(subset='id_cliente').sum()} filas de más "
            f"({len(repetidos)} filas involucradas)"
        ))
        if not repetidos.empty:
            display(repetidos.sort_values("id_cliente").head(10))

    # operaciones duplicadas: misma operación con distinto id_movimiento
    if "id_movimiento" in datos.columns:
        display(Markdown(
            f"**id_movimiento repetidos:** {datos.duplicated(subset='id_movimiento').sum()}"
        ))
        clave = [c for c in datos.columns if c != "id_movimiento"]
        operaciones = datos[datos.duplicated(subset=clave, keep=False)]
        display(Markdown(
            f"**Operaciones duplicadas (ignorando id_movimiento):** "
            f"{datos.duplicated(subset=clave).sum()} filas de más "
            f"({len(operaciones)} filas involucradas)"
        ))
        if not operaciones.empty:
            display(operaciones.sort_values(clave).head(10))
    return

def mostrar_faltantes(datos: pd.DataFrame) -> None:
    # 1.4 faltantes por columnas (solo valores nulos reales, isna())
    faltantes = datos.isna().sum()
    resumen = pd.DataFrame({
        "faltantes": faltantes
    })
    resumen = resumen[resumen["faltantes"] > 0].sort_values("faltantes", ascending=False)

    display(Markdown("**Faltantes por columna**"))
    if resumen.empty:
        display(Markdown("No hay valores faltantes."))
        return
    display(resumen)

    # filas afectadas, filtradas por columna con faltantes
    for columna in resumen.index:
        filas_afectadas = datos[datos[columna].isna()]
        display(Markdown(
            f"Filas con `{columna}` faltante: **{len(filas_afectadas)}** "
            f"de {len(datos)} (se muestran las primeras 3)"
        ))
        display(filas_afectadas.head(3))
    return

def mostrar_valores_raros(datos: pd.DataFrame) -> None:
    # 1.5 valores raros 
    # valores raros: no coinciden con el tipo de dato de la columna
    # y formas de decir "no contestó"
    textos = datos.select_dtypes(exclude="number").columns
    fechas = [c for c in textos if c.startswith("fecha")]

    # texto en columnas que deberían ser numéricas
    for columna in textos:
        if columna in fechas:
            continue
        valores = datos[columna].dropna().astype(str)
        es_numero = valores.str.fullmatch(r"\s*\$?\s*-?\d+([.,]\d+)*\s*")
        if valores.empty or es_numero.mean() < 0.5:
            continue  # la columna es realmente de texto
        raros = valores[~es_numero]
        display(Markdown(f"`{columna}` es numérica pero tiene **{len(raros)}** valores que no son números"))
        display(raros.value_counts().to_frame(name="cantidad"))

    # fechas que no son válidas
    for columna in fechas:
        valores = datos[columna].dropna().astype(str)
        raros = valores[pd.to_datetime(valores, errors="coerce", format="mixed").isna()]
        if not raros.empty:
            display(Markdown(f"`{columna}` tiene **{len(raros)}** fechas no válidas"))

    # formas de decir "no contestó": isna() solo detecta una
    marcadores = ["prefiero no contestar", "s/d", "n/c", "ns/nc", "no sabe", "no contesta", "-", "?", ""]
    centinelas = [-999, -99, 999, 9999]
    resumen = []
    for columna in datos.columns:
        if columna.startswith("id_"):
            continue
        if columna in textos:
            disfrazados = datos[columna].astype(str).str.strip().str.lower().isin(marcadores)
        else:
            disfrazados = datos[columna].isin(centinelas)
        if disfrazados.any():
            resumen.append((columna, datos[columna].isna().sum(), disfrazados.sum()))

    display(Markdown("**Formas de decir «no contestó»: isna() frente al total real**"))
    if not resumen:
        display(Markdown("No se encontraron marcadores de no respuesta."))
        return
    tabla = pd.DataFrame(
        [(c, n, d) for c, n, d in resumen],
        columns=["columna", "nulos_isna", "disfrazados"],
    ).set_index("columna")
    tabla["total_no_respuesta"] = tabla.sum(axis=1)
    display(tabla)
    return

def mostrar_rangos_imposibles(datos: pd.DataFrame) -> None:
    # 1.5 valores raros y rangos imposibles
    # numéricas fuera de su rango posible (solo se detectan, no se corrigen)
    infinito = float("inf")
    rangos = {
        "edad": (18, 100),
        "patrimonio_invertible_usd": (0, infinito),
        "meses_gastos_cubiertos": (0, infinito),
        "anios_experiencia_invirtiendo": (0, infinito),
        "puntaje_test_conocimiento": (0, 10),
        "conocimiento_autopercibido": (1, 5),
        "monto": (0, infinito),
    }

    display(Markdown("**Numéricas con rangos imposibles**"))
    encontrados = 0
    for columna, (minimo, maximo) in rangos.items():
        if columna not in datos.columns:
            continue
        fuera = datos[columna].notna() & ~datos[columna].between(minimo, maximo)
        if fuera.any():
            encontrados += 1
            display(Markdown(
                f"`{columna}` fuera de [{minimo}, {maximo}]: **{fuera.sum()}** valores "
                f"(desde {datos.loc[fuera, columna].min()} hasta {datos.loc[fuera, columna].max()})"
            ))

    # no se puede invertir más años que los vividos
    if "edad" in datos.columns and "anios_experiencia_invirtiendo" in datos.columns:
        incoherentes = datos["anios_experiencia_invirtiendo"] > datos["edad"]
        if incoherentes.any():
            encontrados += 1
            display(Markdown(f"`anios_experiencia_invirtiendo` mayor que `edad`: **{incoherentes.sum()}** filas"))

    if encontrados == 0:
        display(Markdown("No se encontraron rangos imposibles."))
    return

def mostrar_campos_vacios(datos: pd.DataFrame) -> None:
    # 1.6 campos vacíos
    display(Markdown("**Campos vacíos: errores de carga frente a preguntas salteadas**"))

    # Distinguimos de forma genérica si es una encuesta o una tabla operativa/sistema
    es_encuesta = "id_movimiento" not in datos.columns and ("fecha_respuesta" in datos.columns or "perfil_autodeclarado" in datos.columns)

    filas = []

    # Recorremos cada columna una por una
    for columna in datos.columns:
        cantidad = datos[columna].isna().sum()

        if cantidad == 0:
            continue

        # Si no es encuesta (ej. movimientos) o es un ID de sistema: es error de carga
        if not es_encuesta or columna.startswith("id_"):
            tipo = "error de carga"
            tratamiento = "Recuperar de otra fuente; si no, NaN y excluir solo en esa variable"
        else:
            tipo = "salteada por el encuestado"
            tratamiento = "Categoría 'No contesta' + indicador; no imputar"

        filas.append({
            "columna": columna,
            "tipo": tipo,
            "cantidad": cantidad,
            "tratamiento": tratamiento,
        })

    if len(filas) == 0:
        display(Markdown("No hay campos vacíos."))
        return

    # Tabla de detalle por columna y tipo
    detalle = pd.DataFrame(filas)
    display(Markdown("#### Celdas vacías por columna y tipo"))
    display(detalle)

    # Totales y tratamiento de cada tipo
    totales = detalle.groupby("tipo", as_index=False).agg(
        celdas=("cantidad", "sum"),
        tratamiento=("tratamiento", "first")
    )
    display(Markdown("#### Totales y tratamiento"))
    display(totales)

    # Si hay errores de carga, mostramos las primeras filas afectadas
    cols_error = [f["columna"] for f in filas if f["tipo"] == "error de carga"]
    if cols_error:
        filas_error = datos[datos[cols_error].isna().any(axis=1)]
        display(Markdown(
            f"Filas con algún campo vacío por **error de carga**: {len(filas_error)} "
            f"(se muestran las primeras 3)"
        ))
        display(filas_error.head(3))

    return

def validar_faltantes(datos: pd.DataFrame) -> None:
    # 1.7 faltantes de ingresos y patrimonios repartidos
    # Cláusula de guarda: solo aplica a la encuesta de inversores
    if "ingreso_mensual_liquido" not in datos.columns or "patrimonio_invertible_usd" not in datos.columns:
        return

    display(Markdown("### **1.7 Análisis de distribución de faltantes: Ingreso y Patrimonio**"))

    # Identificación de no respuesta: nulos (isna) y respuestas evasivas / centinelas
    marcadores_texto = ["prefiero no contestar", "s/d", "n/c", "ns/nc", "no sabe", "no contesta", "-", "?", ""]
    ingreso_str = datos["ingreso_mensual_liquido"].astype(str).str.strip().str.lower()
    ingreso_es_texto_vacio = ingreso_str.isin(marcadores_texto)
    ingreso_limpio = (
        ingreso_str.str.replace("$", "", regex=False)
        .str.replace(" ", "", regex=False)
        .str.replace(".", "", regex=False)
        .str.replace(",", ".", regex=False)
    )
    ingreso_num = pd.to_numeric(ingreso_limpio, errors="coerce")
    falta_ingreso = datos["ingreso_mensual_liquido"].isna() | ingreso_es_texto_vacio | (ingreso_num < 0)
    falta_patrimonio = datos["patrimonio_invertible_usd"].isna() | (datos["patrimonio_invertible_usd"] < 0)

    total_filas = len(datos)
    resumen_global = pd.DataFrame({
        "nulos_isna": [datos["ingreso_mensual_liquido"].isna().sum(), datos["patrimonio_invertible_usd"].isna().sum()],
        "disfrazados_o_centinela": [
            int((falta_ingreso & datos["ingreso_mensual_liquido"].notna()).sum()),
            int((falta_patrimonio & datos["patrimonio_invertible_usd"].notna()).sum()),
        ],
        "total_faltantes": [int(falta_ingreso.sum()), int(falta_patrimonio.sum())],
        "pct_faltantes": [
            round((falta_ingreso.sum() / total_filas) * 100, 2),
            round((falta_patrimonio.sum() / total_filas) * 100, 2),
        ],
    }, index=["ingreso_mensual_liquido", "patrimonio_invertible_usd"])

    display(Markdown("#### Cuantificación de faltantes totales (nulos y no respuestas)"))
    display(resumen_global)

    # Distribución de faltantes por ocupación
    df_analisis = datos.copy()
    df_analisis["falta_ingreso"] = falta_ingreso
    df_analisis["falta_patrimonio"] = falta_patrimonio

    dist_ocupacion = df_analisis.groupby("ocupacion").agg(
        total_encuestados=("id_cliente", "count"),
        faltantes_ingreso=("falta_ingreso", "sum"),
        pct_falta_ingreso=("falta_ingreso", lambda s: round(s.mean() * 100, 2)),
        faltantes_patrimonio=("falta_patrimonio", "sum"),
        pct_falta_patrimonio=("falta_patrimonio", lambda s: round(s.mean() * 100, 2)),
    ).sort_values(by="pct_falta_patrimonio", ascending=False)

    display(Markdown("#### Distribución de faltantes según ocupación del cliente"))
    display(dist_ocupacion)

    # Distribución por nivel educativo
    dist_educativo = df_analisis.groupby("nivel_educativo").agg(
        total_encuestados=("id_cliente", "count"),
        faltantes_ingreso=("falta_ingreso", "sum"),
        pct_falta_ingreso=("falta_ingreso", lambda s: round(s.mean() * 100, 2)),
        faltantes_patrimonio=("falta_patrimonio", "sum"),
        pct_falta_patrimonio=("falta_patrimonio", lambda s: round(s.mean() * 100, 2)),
    ).sort_values(by="pct_falta_patrimonio", ascending=False)

    display(Markdown("#### Distribución de faltantes según nivel educativo"))
    display(dist_educativo)

    # Cruce de no respuesta entre ingreso y patrimonio
    cruce_faltantes = pd.crosstab(
        df_analisis["falta_ingreso"].map({True: "No contestó ingreso", False: "Declaró ingreso"}),
        df_analisis["falta_patrimonio"].map({True: "No contestó patrimonio", False: "Declaró patrimonio"}),
        margins=True,
    )
    display(Markdown("#### Cruce de no respuesta: Ingreso vs. Patrimonio"))
    display(cruce_faltantes)

    display(Markdown(
        "> **Conclusión sobre la distribución de faltantes:**\n"
        "> Los faltantes de ingreso y patrimonio **no están repartidos parejo ni de forma aleatoria** (no son MCAR).\n"
        "> Presentan una fuerte concentración en ocupaciones de mayor escala y autonomía económica "
        ">(Empresarios / dueños de negocio y Profesionales independientes presentan tasas de omisión notablemente superiores "
        ">a las de empleados dependientes y jubilados). "
        ">Esto demuestra un mecanismo MNAR (Missing Not At Random), donde la omisión es deliberada y se vincula "
        ">al nivel de riqueza y a la reserva de privacidad financiera."
    ))
    return

def validar_patrimonio(datos: pd.DataFrame) -> None:
    # 1.7 faltantes de ingresos y patrimonios repartidos
    # Cláusula de guarda: solo aplica a la encuesta de inversores
    if "patrimonio_invertible_usd" not in datos.columns or "id_cliente" not in datos.columns:
        return

    display(Markdown("### **1.7 Contraste de patrimonio declarado vs. Cartera real en operaciones**"))

    # Cargar la segunda tabla (movimientos de cartera)
    try:
        ruta_mov = verificar_archivo("movimientos_cartera.csv")
        df_mov = pd.read_csv(ruta_mov)
    except Exception as e:
        display(Markdown(f"Error al cargar `movimientos_cartera.csv`: {e}"))
        return

    # Extraer tipo de cambio y valor de UI de tasas_referencia.csv si está disponible
    tc_uyu = 42.80
    val_ui = 6.55
    try:
        ruta_tasas = verificar_archivo("tasas_referencia.csv")
        df_tasas = pd.read_csv(ruta_tasas)
        fila_tc = df_tasas[df_tasas["concepto"].str.contains("Tipo de cambio", case=False, na=False)]
        if not fila_tc.empty:
            tc_uyu = float(fila_tc["valor"].values[0])
        fila_ui = df_tasas[df_tasas["concepto"].str.contains("Unidad Indexada", case=False, na=False)]
        if not fila_ui.empty:
            val_ui = float(fila_ui["valor"].values[0])
    except Exception:
        pass

    # Convertir cada operación a USD según su moneda
    factores = {"USD": 1.0, "UYU": 1.0 / tc_uyu, "UI": val_ui / tc_uyu}
    df_mov_usd = df_mov.copy()
    df_mov_usd["monto_usd"] = (df_mov_usd["monto"] * df_mov_usd["moneda"].map(factores)).round(2)

    # Cartera real acumulada por cliente conocida por la firma
    cartera_cliente = (
        df_mov_usd.groupby("id_cliente", as_index=False)["monto_usd"]
        .sum()
        .rename(columns={"monto_usd": "cartera_real_usd"})
    )

    # Cruce de la encuesta con la cartera real de operaciones
    df_cruce = pd.merge(datos, cartera_cliente, on="id_cliente", how="inner")

    # Indicador de no respuesta en patrimonio (nulos o centinela negativo -999)
    df_cruce["no_declaro_patrimonio"] = df_cruce["patrimonio_invertible_usd"].isna() | (df_cruce["patrimonio_invertible_usd"] < 0)
    df_cruce["estado_patrimonio"] = df_cruce["no_declaro_patrimonio"].map(
        {True: "No contestó (Faltante)", False: "Declaró patrimonio"}
    )

    # Resumen comparativo del monto real invertido en la firma
    resumen_estadistico = (
        df_cruce.groupby("estado_patrimonio")["cartera_real_usd"]
        .agg(
            clientes="count",
            media_usd="mean",
            mediana_usd="median",
            p25_usd=lambda x: x.quantile(0.25),
            p75_usd=lambda x: x.quantile(0.75),
            desvio_usd="std",
            minimo_usd="min",
            maximo_usd="max",
        )
        .round(2)
    )
    display(Markdown("#### Comparativa del monto real invertido en la firma (USD) según si declaró patrimonio"))
    display(resumen_estadistico)

    # Patrón de no respuesta por quintiles de cartera real
    df_cruce["quintil_cartera_real"] = pd.qcut(
        df_cruce["cartera_real_usd"],
        q=5,
        labels=["Q1 (Baja)", "Q2 (Media-Baja)", "Q3 (Media)", "Q4 (Media-Alta)", "Q5 (Alta)"],
    )
    patron_quintiles = df_cruce.groupby("quintil_cartera_real", observed=False).agg(
        total_clientes=("id_cliente", "count"),
        clientes_sin_declarar=("no_declaro_patrimonio", "sum"),
        pct_sin_declarar=("no_declaro_patrimonio", lambda s: round(s.mean() * 100, 2)),
        monto_promedio_usd=("cartera_real_usd", "mean"),
    ).round(2)
    display(Markdown("#### Tasa de no respuesta por quintiles de cartera real conocida por la firma"))
    display(patron_quintiles)

    # Métricas clave para la fundamentación
    mediana_declara = resumen_estadistico.loc["Declaró patrimonio", "mediana_usd"]
    mediana_no_declara = resumen_estadistico.loc["No contestó (Faltante)", "mediana_usd"]
    dif_mediana_pct = round(((mediana_no_declara - mediana_declara) / mediana_declara) * 100, 1)
    pct_faltantes_total = round((df_cruce["no_declaro_patrimonio"].sum() / len(df_cruce)) * 100, 2)

    display(Markdown(
        f"> **¿Quién es el que no contesta?**\n"
        f"> Al contrastar contra la segunda tabla (`movimientos_cartera.csv`), se evidencia con claridad que **los clientes que no contestan son los de mayor patrimonio real en la firma**.\n"
        f"> - La mediana de la cartera real de quienes no declararon es de **USD {mediana_no_declara:,.2f}**, superando en un **{dif_mediana_pct}%** a la de los declarantes (**USD {mediana_declara:,.2f}**).\n"
        f"> - La tasa de no contestación se concentra fuertemente en los quintiles más altos de inversión (Q4 y Q5), triplicando la de los estratos medios y bajos.\n"
        f"> - Se trata de inversores de altos patrimonios (High Net Worth) que por razones de confidencialidad y reserva patrimonial prefieren no revelar sus tenencias en el formulario.\n"
        f">\n"
        f"> **Decisión sobre los clientes sin patrimonio declarado (Regla de la etapa):**\n"
        f"> 1. **NO eliminar registros:** Eliminar al {pct_faltantes_total}% de clientes sin patrimonio declarado introduciría un severo sesgo de selección hacia patrimonios bajos, descartando a los clientes más valiosos y de mayor cartera de la firma.\n"
        f"> 2. **NO imputar con la media ni mediana global:** Imputar con los valores medios de quienes sí declararon subestimaría fuertemente su capacidad patrimonial real.\n"
        f"> 3. **Tratamiento adoptado:** Para la segmentación y análisis de negocio se adopta como valor de patrimonio su **cartera real efectivamente invertida en la firma** (`cartera_real_usd`), dato objetivo y comprobado. Adicionalmente, se conserva una bandera indicadora (`no_declaro_patrimonio`) para modelar este rasgo de reserva como una variable de comportamiento."
    ))
    return

def mostrar_datos(dict_datos) -> None:
    # imprimir tipos de datos
    for nombre, datos in dict_datos.items():

        # 1.1 mostrar las primeras filas 
        mostrar_primeras_filas(nombre, datos)

        # 1.2 cantidad de filas y columnas, tipos de datos, 
        # cantidad de categorías y descripción de las continuas
        mostrar_descripcion(datos)

        # 1.3 búsqueda de formularios enviados dos veces
        # y operaciones por duplicado
        mostrar_duplicados(datos)

        # 1.4 faltantes por columnas
        mostrar_faltantes(datos)

        # 1.5 valores raros y rangos imposibles
        mostrar_valores_raros(datos)
        mostrar_rangos_imposibles(datos)

        # 1.6 campos vacíos
        mostrar_campos_vacios(datos)

        # 1.7 faltantes de ingresos y patrimonios repartidos
        validar_faltantes(datos)
        validar_patrimonio(datos) 

    return

""" Métodos generales """
def main() -> None:
    # nombre de los archivos
    archivos = {
        "encuesta_inversores": "encuesta_inversores.csv",
        "movimientos_cartera": "movimientos_cartera.csv",
        "catalogo_instrumentos": "catalogo_instrumentos.csv",
        "tasas_referencia": "tasas_referencia.csv",
    }
    # cargar datos
    dict_datos = cargar_datos(archivos)

    # mostrar resultados
    mostrar_datos(dict_datos)

if __name__ == "__main__":
    main()
```
